```python
#sorted
with open("f1.txt","r") as file:
  data=file.read().lower()
words=data.split()
words.sort()
with open("f2.txt","w")as file:
  for word in words:
    file.write(word+"\n")
```

```python
#python program to compute the no of characters,words and lines in a file
with open("f1.txt","r") as file:
  data=file.read()
  lines=data.split("\n")
  words=data.split()
  characters=list(data)
  print("lines:",len(lines))
  print("words:",len(words))
  print("characters:",len(characters))
```
