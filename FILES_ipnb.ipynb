```python
#program on write mode
file=open("f1.txt","w")
data="hello world"
file.write(data)
print("file created")
file.close()

```

```python
#program on r mode
with open("f1.txt","r") as f:
  data=f.read()
  print(data)
```

```python
with open("f1.txt","r") as f:
  print(f.readline())
  print(f.read())
  print(f.readlines())
```

```python
with open("f1.txt","r") as f:
  print(f.readline())
  print(f.tell())
  f.seek(1)
  print(f.tell())
  print(f.read())
```

```python
#copy a file
with open("f1.txt","r") as f:
  data=f.read()
print("reading done")
with open("f2.txt","w") as f2:
  f2.write(data)
print("writing done")
```

```python

#program to print reverse of a line
with open("f1.txt","r") as f:
  data=f.readlines()
for i in data:
  print(i[: :-1])

```

```python
#program to sort words in a file and put them in another file.The output should be in lower case
with open("f1.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f3.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
  print("done")
```

```python
#python program to compute the number of characters,words in a file
f=open( "f1.txt","r")
data=f.read()
print("number of characters",len(data))
print("number of words",len(data.split()))
f.close()
```
