```python
# dictionary comprehension
d={i:i*i for i in range(1,6)}
print(d)
```

```python
#user defined dictionary
k=[i for i in input().split()]
v=[int(i) for i in input().split()]
d={i:j for i,j in zip(k,v)}
print(d)

```

```python
# dictionary operations
d={"name":"lakshmi","age":18,"branch":"AI&DS"}
#access
print("name:",d["name"])
#update
d["age"]=19
print("after updating:",d)
#delete
del d["branch"]
print("after deleting:",d)
#membership
print("name" in d)
#length
print("length:",len(d))
#iteration
print("dictionary elements:")
for key in d:
  print(key,":",d[key])
```

```python
# write a program to check key exits in a dictionary
d={"sno":210,"name":"sai","grade":8.9}
key=input()
if key in d:
  print("yes")
else:
  print("no")
```

```python
#write a program to add new pair in the existing dictionary
d={"sno":210,"name":"sai","grade":8.9}
print("before adding:",d)
d["age"]=18
print("after adding:",d)
```

```python
#write a python program to sum all the item in a given dictionary
d={'a':12,'b':34,'c':56,'d':78}
sum=sum(d.values())
print(sum)
```

```python
# write a program that prints frequency of each element in a list in following formate
#input:[1,2,3,4,1,2,3,4,5,4,6,6]   output:{1:2,2:2,3:2,4:3,5:1,6:2}
l=[1,2,3,4,1,2,3,4,5,4,6,6]    #l=[int(i) for i in input().split()] input from user
d={}
for i in l:
  d[i]=l.count(i)
print(d)
```
