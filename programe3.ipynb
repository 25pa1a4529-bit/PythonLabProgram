```python
#1.list operations
l1=[21,3,'lakshmi']
l2=[28,2,'h']

#concatination
print(l1+l2)

#repeatation
print(l1*4)

#indexing
print(l1[1])

#slicing
print(l1[1:])

#membership operator
if 21 in l1:
  print("YES")
else:
  print("NO")

#iteration
for i in l2:
  print(i)
```

```python
#2.builtin functions of lists
l1=[21,3,28,2]

#len()
print(len(l1))

#sum
print(sum(l1))

#min()
print(min(l1))

#max()
print(max(l1))

#reversed()
print(list(reversed(l1)))

#sorted()
print(sorted(l1))
```

```python
#3.builtin methods of lists
l1=[21,3,"l"]
l2=[28,2,"h"]

#append
l1.append("k")
print(l1)

#extent
l2.extend(l1)
print(l2)

#insert
l2.insert(2,"l")
print(l2)

#pop
l2.pop(2)
print(l2)

#remove
l2.remove(28)
print(l2)

#count
print(l1.count(3))

#reverse
l1.reverse()
print(l1)

#sort
l3=[3,545,54,4185,15,456,296,8]
l3.sort()
print(l3)
```

```python
#program that prints sum of list elements without using sum()
l1=[4,5,8,34]
sum=0
for i in l1:
  sum=sum+i
print(sum)
```

```python
#print list elements at odd index
l1=[4,8,5,6,45,525]
for i in range(0,len(l1),2):
  print(l1[i])
```

```python
#print the even elements
l1=[25,74,56,9,2,8,5,4]
for i in l1:
  if i%2==0:
    print(i)
```

```python
#program on list comprehension
l=[for i in input().split() if i%2==0]

print(f"the even numbers in a given range are: {l}")
```
