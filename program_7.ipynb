```python
#set operations
s1={1,3,5,7,"hey"}
s2={2,4,6,8}
#union
print(s1.union(s2))
#intersection
print(s1.intersection(s2))
#difference
print(s1.difference(s2))
#symmetric difference
print(s1.symmetric_difference(s2))
```

```python
#set methods
s1={1,3,5,7,"hey"}
s2={2,4,6,8}
#add
s1.add(9)
print(s1)
#update
s1.update([11,3])
print(s1)
#remove
s1.remove(3)
print(s1)
#discard
s1.discard(5)
print(s1)
#pop
s1.pop()
print(s1)
#clear
s1.clear()
print(s1)
#union
print(s1.union(s2))
#intersection
print(s1.intersection(s2))
#difference
print(s1.difference(s2))
#symmetric difference
print(s1.symmetric_difference(s2))
```
