```python
#this code is for arthematic operations
a=int(input("enter a number"))
b=int(input("enter another number"))
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
```

```python
#this code is for comparision operations
a=int(input("enter a number"))
b=int(input("enter another number"))
print(a==b)
print(a!=b)
print(a<b)
print(a>b)
print(a<=b)
print(a>=b)
```

```python
#this code is for logical operations
a=int(input("enter a number"))
b=int(input("enter another number"))
print(a!=b and a<b)
print(a<b or a==b)
print(not(a<b))

```

```python
#this code is for assignment operations
x=int(input("enter a number"))
print(x)
x += 5
print(x)
x -=3
print(x)
x *= 2
print(x)
x /= 4
print(x)
```

```python
#this code is for identify operations
x = [1, 2]
y = x
z = [1, 2]
print(x is y)
print(x is z)
print(x is not z)
```

```python
#this code is for membership operations
fruits = ["apple", "banana", "cherry"]
print("banana" in fruits)
print("orange" in fruits)
print("orange" not in fruits)
```

```python
#this code is for bitwise operations
a=int(input("enter a number"))
b=int(input("enter another number"))
print(a & b)
print(a | b)
print(a ^ b)
```

```python
#this code is for terinary operations
age = int(input())
status = "Adult" if age >= 18 else "Minor"
print(status)
```
