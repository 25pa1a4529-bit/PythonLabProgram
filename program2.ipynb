```python
#program that finds sum of digits of a number
#56=5+6=11
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```

```python
#reverse of a number
n=int(input())
rev=0
while(n!=0):
  rem=n%10
  rev=rev*10 +rem
  n=n//10
print(rev)
```

```python
#check a given number is prime or not
n=int(input())
for i in range(2,n):
  if(n%i==0):
    print("not a prime num")
    break
else:
    print("it is  a prime")
```
